# Imports

```python
import pandas as pd
```

```python
player_boxscores = pd.read_csv("data/player_boxscores.csv")
```

```python
def get_joining_dates_for_player(player_id: int):
    # GET TEAMS WHERE THE PLAYER PLAYED
    teams_where_player_played = player_boxscores[player_boxscores["player_id"] == player_id][["team_id", "team_abbreviation", "team_name"]].drop_duplicates()

    for idx, team in teams_where_player_played.iterrows():
        # PLAYER NAME AND ID
        teams_where_player_played.loc[idx, "player_id"] = player_id
        teams_where_player_played.loc[idx, "player_name"] = player_boxscores[
            (player_boxscores["player_id"] == player_id) &
            (player_boxscores["team_abbreviation"] == team["team_abbreviation"])
        ]["player_name"].iloc[0]

        # JOINING DATE
        date_of_joining = player_boxscores[
            (player_boxscores["player_id"] == player_id) &
            (player_boxscores["team_abbreviation"] == team["team_abbreviation"])
        ]["game_date"].min()
        teams_where_player_played.loc[idx, "date_of_joining"] = date_of_joining

        # EXIT DATE
        date_of_exit = player_boxscores[
            (player_boxscores["player_id"] == player_id) &
            (player_boxscores["team_abbreviation"] == team["team_abbreviation"])
        ]["game_date"].max()
        teams_where_player_played.loc[idx, "date_of_exit"] = date_of_exit

    return teams_where_player_played

def get_joining_dates_for_players(players_id: list[int]):
    res_df = pd.DataFrame()
    for id in players_id:
        res_df = pd.concat([res_df, get_joining_dates_for_player(id)])

    return res_df    
```

```python
# Usage
lebron_id = player_boxscores[player_boxscores["player_name"] == "LeBron James"]["player_id"].iloc[0]
shaquille_id = player_boxscores[player_boxscores["player_name"] == "Shaquille O'Neal"]["player_id"].iloc[0]

print(f"LeBron James ID: {lebron_id}")
print(f"Shaquille O'Neal ID: {shaquille_id}")

display(get_joining_dates_for_player(lebron_id))
display(get_joining_dates_for_players([lebron_id, shaquille_id]))
```
